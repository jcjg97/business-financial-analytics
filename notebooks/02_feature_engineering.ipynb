# Feature Engineering for Financial Analytics

This notebook demonstrates professional feature engineering and data preparation for a financial dataset. The goal is to create a clean, enriched dataset ready for machine learning and business intelligence (BI) applications.

## 1. Load Dataset

```python
import pandas as pd
import numpy as np

# Load the dataset
data_path = "../data/raw/financial_data.csv"
df = pd.read_csv(data_path)
df.head()
```

## 2. Convert Date Column to Datetime

```python
# Convert 'date' column to datetime
df['date'] = pd.to_datetime(df['date'])
df.dtypes
```

## 3. Handle Missing Values

```python
# Check for missing values
missing = df.isnull().sum()
print('Missing values per column:')
print(missing)

# Example: Fill missing numerical values with median, categorical with mode
for col in df.columns:
    if df[col].dtype in [np.float64, np.int64]:
        df[col] = df[col].fillna(df[col].median())
    elif df[col].dtype == 'object':
        df[col] = df[col].fillna(df[col].mode()[0])

# Verify no missing values remain
df.isnull().sum()
```

## 4. Feature Engineering: Profit, Margin, Revenue per Unit

```python
# Create new features
df['profit'] = df['revenue'] - df['cost']
df['margin'] = df['profit'] / df['revenue']
df['revenue_per_unit'] = df['revenue'] / df['units_sold']
df[['profit', 'margin', 'revenue_per_unit']].head()
```

## 5. Extract Time-Based Features

```python
# Extract year and month
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df[['date', 'year', 'month']].head()
```

## 6. Encode Categorical Variables

```python
# One-hot encode 'region' and 'product'
df = pd.get_dummies(df, columns=['region', 'product'], drop_first=False)
df.head()
```

## 7. Remove Duplicates

```python
# Remove duplicate rows if any
before = df.shape[0]
df = df.drop_duplicates()
after = df.shape[0]
print(f'Removed {before - after} duplicate rows.')
```

## 8. Prepare Final Cleaned Dataset

```python
# Preview the cleaned dataset
df.head()

# Export to CSV con formato correcto para números
output_path = "../data/processed/financial_data_clean.csv"
df.to_csv(output_path, index=False, sep=';', decimal=',', float_format='%.2f')
print(f'Cleaned dataset saved to: {output_path}')
```
